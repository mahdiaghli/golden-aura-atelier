```python
import requests
import pandas as pd


# -----------------------------
# تنظیمات
# -----------------------------

BASE_URL = "http://94.183.48.41:8090"

USERNAME = "manager"
PASSWORD = "123456"

BUSINESS_ID = 4

PAGE_COUNT = 500


# -----------------------------
# مرحله اول: Login
# -----------------------------

print("Logging in...")

login_response = requests.get(
    f"{BASE_URL}/Services/Login/",
    params={
        "username": USERNAME,
        "password": PASSWORD
    }
)

login_data = login_response.json()

if login_data["Status"] != "OK":
    raise Exception(login_data)


USERKEY = login_data["Result"]["UserKey"]

print("Login موفق")
print("UserKey:", USERKEY)


# -----------------------------
# مرحله دوم: دریافت محصولات
# -----------------------------

products = []

page_index = 1


while True:

    print(f"دریافت صفحه {page_index}")

    response = requests.get(
        f"{BASE_URL}/Services/MadeGold/List/",
        params={
            "userkey": USERKEY,
            "BusinessId": BUSINESS_ID,
            "PageIndex": page_index,
            "PageCount": PAGE_COUNT
        }
    )


    data = response.json()


    if data["Status"] != "OK":
        print(data)
        break


    result = data["Result"]


    # اگر صفحه خالی بود یعنی تمام شده
    if not result:
        break


    for p in result:

        products.append({

            "ProductId": p.get("ProductId"),

            "کد محصول": p.get("ProductCode"),

            "نام محصول": p.get("ProductTitle"),

            "وزن": p.get("Weight"),

            "وزن ترازو": p.get("BaseWeight"),

            "عیار": p.get("Carat"),

            "سنگ ها": p.get("StonesNames"),

            "وضعیت موجودی": p.get("IsExists"),

            "دسته بندی": p.get("CategoryTitle"),

            "ویترین": p.get("LocationTitle"),


            "تصویر1":
                BASE_URL + p["ImageURL1"]
                if p.get("ImageURL1") else "",

            "تصویر2":
                BASE_URL + p["ImageURL2"]
                if p.get("ImageURL2") else "",

            "تصویر3":
                BASE_URL + p["ImageURL3"]
                if p.get("ImageURL3") else "",

            "تصویر4":
                BASE_URL + p["ImageURL4"]
                if p.get("ImageURL4") else "",

            "تصویر5":
                BASE_URL + p["ImageURL5"]
                if p.get("ImageURL5") else "",

            "تصویر6":
                BASE_URL + p["ImageURL6"]
                if p.get("ImageURL6") else "",


            "مدل": p.get("ModelTitle"),

            "برند": p.get("BrandTitle"),

            "رنگ": p.get("ColorTitle"),

            "سایز": p.get("SizeTitle"),


            "قیمت طلا": p.get("GoldPrice"),

            "قیمت سنگ": p.get("StonePrice"),

            "اجرت": p.get("WageOfPrice"),

            "سود": p.get("IncomeTotal"),

            "مالیات": p.get("TaxTotal"),

            "قیمت نهایی": p.get("TotalPrice"),


            "کالکشن": p.get("CollectionTitle"),

            "طراح": p.get("Designer"),

            "کد طراحی": p.get("DesignerCode"),

            "کد قدیمی": p.get("OldCode"),

            "کد دفتری": p.get("OfficeCode"),


            "رزرو شده":
                p.get("IsReserved")
        })


    page_index += 1



# -----------------------------
# ذخیره Excel
# -----------------------------

df = pd.DataFrame(products)

df.to_excel(
    "MadeGold_products.xlsx",
    index=False
)


print("==========================")
print("تعداد محصولات:", len(products))
print("فایل ساخته شد: MadeGold_products.xlsx")
print("==========================")
```

```python
import os
import re
import base64
from openpyxl import load_workbook

# -----------------------------
# تنظیمات
# -----------------------------

EXCEL_FILE = "accountinggoldproductsearchprint.xlsx"
IMAGE_FOLDER = "ExtractedImages"

# -----------------------------
# خواندن فایل
# -----------------------------

wb = load_workbook(EXCEL_FILE)
ws = wb.active

# -----------------------------
# پیدا کردن ستون تصویر
# -----------------------------

image_col = None

for cell in ws[1]:
    if cell.value is not None and "تصویر" in str(cell.value):
        image_col = cell.column_letter
        break

if image_col is None:
    raise Exception("ستون تصویر پیدا نشد.")

print(f"ستون تصویر: {image_col}")
print("=" * 70)

# -----------------------------
# بررسی محتوای سلول ها
# -----------------------------

url_pattern = re.compile(r"^https?://", re.IGNORECASE)

os.makedirs(IMAGE_FOLDER, exist_ok=True)

for row in range(2, ws.max_row + 1):

    value = ws[f"{image_col}{row}"].value

    print(f"ردیف {row}:")

    if value is None or str(value).strip() == "":
        print("   ➜ خالی")

    elif isinstance(value, str):

        text = value.strip()

        # URL
        if url_pattern.match(text):
            print("   ➜ URL")
            print("   ", text)

        # Base64
        elif text.startswith("data:image") or len(text) > 300:

            try:

                if "," in text:
                    text = text.split(",", 1)[1]

                image_data = base64.b64decode(text)

                filename = os.path.join(
                    IMAGE_FOLDER,
                    f"base64_row_{row}.png"
                )

                with open(filename, "wb") as f:
                    f.write(image_data)

                print("   ➜ Base64 Image")
                print("   ذخیره شد:", filename)

            except Exception:
                print("   ➜ متن طولانی (Base64 معتبر نیست)")

        # مسیر فایل
        elif os.path.exists(text):

            print("   ➜ مسیر فایل")
            print("   ", text)

        else:

            print("   ➜ متن")
            print("   ", text)

    else:

        print("   ➜ نوع داده:", type(value))
        print("   ", value)

    print("-" * 70)

# -----------------------------
# استخراج تصاویر Embed شده
# -----------------------------

print("\nبررسی تصاویر Embed شده...")
print("=" * 70)

if hasattr(ws, "_images") and len(ws._images) > 0:

    print(f"{len(ws._images)} تصویر پیدا شد.\n")

    for i, img in enumerate(ws._images):

        filename = os.path.join(
            IMAGE_FOLDER,
            f"embedded_{i+1}.png"
        )

        with open(filename, "wb") as f:
            f.write(img._data())

        print("ذخیره شد:", filename)

else:

    print("هیچ تصویر Embed شده‌ای داخل فایل وجود ندارد.")
```
