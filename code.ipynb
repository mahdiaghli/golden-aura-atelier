```python
import requests
import pandas as pd


# -----------------------------
# تنظیمات
# -----------------------------

BASE_URL = "http://94.183.48.41:8090"

USERNAME = "manager"
PASSWORD = "123456"

BUSINESS_ID = 4

PAGE_COUNT = 500


# -----------------------------
# مرحله اول: Login
# -----------------------------

print("Logging in...")

login_response = requests.get(
    f"{BASE_URL}/Services/Login/",
    params={
        "username": USERNAME,
        "password": PASSWORD
    }
)

login_data = login_response.json()

if login_data["Status"] != "OK":
    raise Exception(login_data)


USERKEY = login_data["Result"]["UserKey"]

print("Login موفق")
print("UserKey:", USERKEY)


# -----------------------------
# مرحله دوم: دریافت محصولات
# -----------------------------

products = []

page_index = 1


while True:

    print(f"دریافت صفحه {page_index}")

    response = requests.get(
        f"{BASE_URL}/Services/MadeGold/List/",
        params={
            "userkey": USERKEY,
            "BusinessId": BUSINESS_ID,
            "PageIndex": page_index,
            "PageCount": PAGE_COUNT
        }
    )


    data = response.json()


    if data["Status"] != "OK":
        print(data)
        break


    result = data["Result"]


    # اگر صفحه خالی بود یعنی تمام شده
    if not result:
        break


    for p in result:

        products.append({

            "ProductId": p.get("ProductId"),

            "کد محصول": p.get("ProductCode"),

            "نام محصول": p.get("ProductTitle"),

            "وزن": p.get("Weight"),

            "وزن ترازو": p.get("BaseWeight"),

            "عیار": p.get("Carat"),

            "سنگ ها": p.get("StonesNames"),

            "وضعیت موجودی": p.get("IsExists"),

            "دسته بندی": p.get("CategoryTitle"),

            "ویترین": p.get("LocationTitle"),


            "تصویر1":
                BASE_URL + p["ImageURL1"]
                if p.get("ImageURL1") else "",

            "تصویر2":
                BASE_URL + p["ImageURL2"]
                if p.get("ImageURL2") else "",

            "تصویر3":
                BASE_URL + p["ImageURL3"]
                if p.get("ImageURL3") else "",

            "تصویر4":
                BASE_URL + p["ImageURL4"]
                if p.get("ImageURL4") else "",

            "تصویر5":
                BASE_URL + p["ImageURL5"]
                if p.get("ImageURL5") else "",

            "تصویر6":
                BASE_URL + p["ImageURL6"]
                if p.get("ImageURL6") else "",


            "مدل": p.get("ModelTitle"),

            "برند": p.get("BrandTitle"),

            "رنگ": p.get("ColorTitle"),

            "سایز": p.get("SizeTitle"),


            "قیمت طلا": p.get("GoldPrice"),

            "قیمت سنگ": p.get("StonePrice"),

            "اجرت": p.get("WageOfPrice"),

            "سود": p.get("IncomeTotal"),

            "مالیات": p.get("TaxTotal"),

            "قیمت نهایی": p.get("TotalPrice"),


            "کالکشن": p.get("CollectionTitle"),

            "طراح": p.get("Designer"),

            "کد طراحی": p.get("DesignerCode"),

            "کد قدیمی": p.get("OldCode"),

            "کد دفتری": p.get("OfficeCode"),


            "رزرو شده":
                p.get("IsReserved")
        })


    page_index += 1



# -----------------------------
# ذخیره Excel
# -----------------------------

df = pd.DataFrame(products)

df.to_excel(
    "MadeGold_products.xlsx",
    index=False
)


print("==========================")
print("تعداد محصولات:", len(products))
print("فایل ساخته شد: MadeGold_products.xlsx")
print("==========================")
```

```python
import pandas as pd
import json
import re

# ==================== تنظیمات ====================
excel_path = "accountinggold_with_image_names.xlsx"   # مسیر فایل اکسل
output_json = "products.json"                        # نام فایل خروجی
sheet_name = 0                                       # شماره یا نام شیت

# ==================== خواندن اکسل ====================
df = pd.read_excel(excel_path, sheet_name=sheet_name)

# حذف ردیف‌های خالی و ردیف توضیحات
df = df.dropna(how="all")
df = df[df.iloc[:, 0].notna()]          # فقط ردیف‌هایی که شماره ردیف دارند
df = df[~df.iloc[:, 0].astype(str).str.contains("توضیحات", na=False)]

# ==================== مپ کردن ستون‌ها ====================
# ستون‌های اکسل شما:
# ردیف | دسته بندی | محصول | وزن | عیار | فی تهیه اجرت درصدی | اجرت درصدی فروش | ... | رنگ | نوع کالا | سایز | کد قدیمی | تصویر

def extract_code(product_name):
    """استخراج کد محصول از ستون محصول (مثل GD01012636)"""
    if pd.isna(product_name):
        return None
    match = re.search(r"(GD\d+)", str(product_name), re.IGNORECASE)
    return match.group(1).upper() if match else None

def map_category(persian_type):
    """تبدیل نوع کالا به category انگلیسی"""
    mapping = {
        "النگو": "bracelets",
        "دستبند": "bracelets",
        "گوشواره": "earrings",
        "انگشتر": "rings",
        "مدال": "necklaces",
        "تک پوش": "necklaces",
        "گردنبند": "necklaces",
        "زنجیر": "necklaces",
        "آویز": "necklaces",
    }
    return mapping.get(str(persian_type).strip(), "other")

def map_type_label(persian_type):
    """تبدیل نوع کالا به typeLabel انگلیسی"""
    mapping = {
        "النگو": "Bangle",
        "دستبند": "Bracelet",
        "گوشواره": "Earrings",
        "انگشتر": "Ring",
        "مدال": "Pendant",
        "تک پوش": "Single Pendant",
        "گردنبند": "Necklace",
        "زنجیر": "Chain",
        "آویز": "Pendant",
    }
    return mapping.get(str(persian_type).strip(), str(persian_type))

def map_color(persian_color):
    """تبدیل رنگ به انگلیسی"""
    mapping = {
        "زرد": "yellow",
        "سفید": "white",
        "دو رنگ": "two-tone",
        "دورنگ": "two-tone",
        "رزگلد": "rose",
        "رز گلد": "rose",
    }
    return mapping.get(str(persian_color).strip(), str(persian_color).lower() if persian_color else None)

def map_gender(category):
    """تقریبی جنسیت بر اساس نوع کالا"""
    if category in ["earrings", "bracelets"] and "بangle" not in category:
        return "women"
    return "unisex"

# ==================== ساخت لیست محصولات ====================
products = []

for _, row in df.iterrows():
    product_name = row.get("محصول") or row.iloc[2]
    code = extract_code(product_name)
    
    if not code:
        continue

    weight = row.get("وزن") or row.iloc[3]
    karat = row.get("عیار") or row.iloc[4]
    making_pct_raw = row.get("اجرت درصدی فروش") or row.iloc[6]   # اجرت درصدی فروش
    color_fa = row.get("رنگ") or row.iloc[11]
    type_fa = row.get("نوع کالا") or row.iloc[12]
    size = row.get("سایز") or row.iloc[13]
    image = row.get("تصویر") or row.iloc[15]

    # تبدیل عیار (750 → 18K)
    karat_str = "18K" if str(karat).strip() in ["750", "18", "18K"] else f"{karat}K"

    # تبدیل درصد اجرت (مثلاً 7 → 0.07)
    try:
        making_pct = float(making_pct_raw) / 100 if making_pct_raw and float(making_pct_raw) > 1 else float(making_pct_raw or 0)
    except:
        making_pct = 0.0

    category = map_category(type_fa)
    type_label = map_type_label(type_fa)
    color = map_color(color_fa)

    # ساخت نام محصول
    name = f"{type_label} {code}"

    item = {
        "id": code.lower(),
        "name": name,
        "category": category,
        "karat": karat_str,
        "weight": float(weight) if pd.notna(weight) else 0,
        "makingPct": round(making_pct, 4),
        "gender": map_gender(category),
        "color": color,
        "code": code,
        "typeLabel": type_label,
    }

    # فقط اگر سایز وجود داشت اضافه کن
    if pd.notna(size) and str(size).strip() not in ["", "nan", "None"]:
        item["size"] = str(size).strip()

    # فقط اگر تصویر واقعی وجود داشت اضافه کن
    if pd.notna(image) and str(image).strip() not in ["", "nan", "None", "nopicture.png"]:
        item["imageName"] = str(image).strip()

    products.append(item)

# ==================== ذخیره JSON ====================
with open(output_json, "w", encoding="utf-8") as f:
    json.dump(products, f, ensure_ascii=False, indent=2)

print(f"✅ تعداد {len(products)} محصول با موفقیت به فایل {output_json} تبدیل شد.")
```

```python
import pandas as pd

excel_path = "aaa.xlsx"
output_csv = "phones.csv"
phone_columns = ["موبایل", "تلفن"]   # ستون‌هایی که می‌خوایم پردازش کنیم

# مهم: dtype=str تا اعشاری نشن و صفر اول حذف نشه
df = pd.read_excel(excel_path, dtype=str)

print("ستون‌های موجود:")
print(df.columns.tolist())

def clean_and_add_zero(series):
    series = (
        series.astype(str)
        .str.strip()
        .str.replace(r"\.0$", "", regex=True)   # حذف .0 احتمالی باقی‌مانده از عدد اعشاری
        .str.replace(r"\s+", "", regex=True)    # حذف فاصله‌های داخلی
    )
    series = series.apply(lambda x: x if x.startswith("0") else "0" + x)
    return series

# فقط ستون‌هایی که واقعاً در فایل وجود دارن رو پردازش کن
found_columns = [c for c in phone_columns if c in df.columns]
if not found_columns:
    raise ValueError("هیچ‌کدام از ستون‌های شماره تلفن پیدا نشد")

# همه شماره‌ها رو (موبایل + تلفن) توی یک ستون جمع می‌کنیم
all_phones = pd.concat([df[c].dropna() for c in found_columns], ignore_index=True)
all_phones = clean_and_add_zero(all_phones)

result = pd.DataFrame({"شماره": all_phones})
result.to_csv(output_csv, index=False, encoding="utf-8-sig")
print(f"✅ {len(result)} شماره در {output_csv} ذخیره شد.")
```

```python
import pandas as pd

# فایل ورودی
input_file = "phones.csv"

# فایل خروجی
output_file = "phones_aaa_fixed.csv"

# خواندن فایل
df = pd.read_csv(input_file, dtype=str)

# نام ستون شماره تلفن
phone_column = "شماره"  # اگر نام ستون فرق دارد، این را تغییر بده

# تبدیل شماره‌ها
df[phone_column] = (
    df[phone_column]
    .fillna("")
    .str.strip()
    .str.replace(r"^0", "+98", regex=True)
)

# ذخیره فایل جدید
df.to_csv(output_file, index=False, encoding="utf-8-sig")

print(f"تمام شد. فایل جدید ذخیره شد: {output_file}")
```

```python
import pandas as pd

# فایل ورودی
csv_file = "phones_bbb_fixed.csv"

# فایل خروجی
vcf_file = "contacts.vcf"

# خواندن CSV
df = pd.read_csv(csv_file, dtype=str)

# نام ستون شماره تلفن
phone_column = "شماره تماس"  # اگر نام ستون فرق دارد، تغییرش بده

with open(vcf_file, "w", encoding="utf-8") as f:
    for i, phone in enumerate(df[phone_column].fillna("")):
        phone = phone.strip()

        if not phone:
            continue

        name = f"Contact {i+1}"

        f.write("BEGIN:VCARD\n")
        f.write("VERSION:3.0\n")
        f.write(f"FN:{name}\n")
        f.write(f"TEL;TYPE=CELL:{phone}\n")
        f.write("END:VCARD\n")

print(f"فایل {vcf_file} ساخته شد.")
```
